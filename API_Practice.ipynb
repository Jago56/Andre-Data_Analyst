 GET request example using Python’s requests library. It shows how to call an API, check the response, and work with the returned JSON.
 What this example demonstrates

Call an API using a GET request

Check the status code

Convert the response into JSON

Print and inspect the returned data

```python
import requests

# API endpoint
url = "https://jsonplaceholder.typicode.com/posts/1"

# Send GET request
response = requests.get(url)

# Check status code
print("Status Code:", response.status_code)

# Convert response to JSON
data = response.json()

# Display the JSON data
print("Response JSON:", data)
```

```python
import requests

url = "https://jsonplaceholder.typicode.com/comments"
params = {"postId": 1}

response = requests.get(url, params=params)
comments = response.json()

for c in comments[:3]:
    print(c)
```

POST request example using Python’s requests library. Shows how to send data to an API, receive a JSON response, and inspect the results.

```python
import requests
import json

url = "https://jsonplaceholder.typicode.com/posts"

payload = {
    "title": "API Demo",
    "body": "Testing POST request from Colab",
    "userId": 99
}

response = requests.post(url, json=payload)

print("Status:", response.status_code)
print("Response JSON:", response.json())
```

Parsing JSON Into Pandas DataFrame

```python
import requests
import pandas as pd

url = "https://jsonplaceholder.typicode.com/posts"
response = requests.get(url)

data = response.json()

df = pd.DataFrame(data)
df.head()
```

API With Headers

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts/1" # Changed to a valid API endpoint
headers = {
    # "Authorization": "Bearer YOUR_API_KEY", # Removed as not needed for this public API
    "Accept": "application/json"
}

response = requests.get(url, headers=headers)
data = response.json()

print(data)
```

Error Handling (Best Practice)

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts/1"

try:
    response = requests.get(url)
    response.raise_for_status()  # throws error for bad status codes
    data = response.json()
    print(data)

except requests.exceptions.RequestException as e:
    print("API Error:", e)
```
