# Bar Plot
- Bar plot used to create the graphics for x and y axis to visualize the categorical data with respect to numerical values in form of a rectangular bar.

```python
# dummy data
x = ['Python', 'MySql', 'C++','Java', 'PowerBI', 'Machine L', 'AI']

y = [500, 550, 450, 650, 800, 1500, 1900]
```

```python
#create a simple bar plot
plt.bar(x, y)

# function to show the plt

plt.show()

# download the figure
plt.savefig('bar.png')
```

# labeling of the plot

```python
bar = plt.bar_label(plt.bar(x,y))

#title
plt.title('Course Price Distribution')

#x axis
plt.xlabel('Courses Name')

#y Axis
plt.ylabel('Fees in $')

plt.show()
```

```python
# increase the size of lable and title
bar = plt.bar_label(plt.bar(x,y))

#title
plt.title('Course Price Distribution', fontsize = 15)

#x axis
plt.xlabel('Courses Name', fontsize = 13)

#y Axis
plt.ylabel('Fees in $', fontsize = 13)

plt.show()
```

changing colours font for title and labels

```python
bar = plt.bar_label(plt.bar(x,y))

#title
plt.title('Course Price Distribution', fontsize = 15, color = 'g')

#x axis
plt.xlabel('Courses Name', fontsize = 13, color = 'r')

#y Axis
plt.ylabel('Fees in $', fontsize = 13, color = 'b')

plt.show()
```

# Line Chart

# Creating the line chart with two parameters

# if the temperature increase the sales of cold drinks and ice cream also increases

```python
import matplotlib.pyplot as plt
# create a line plot
month = ['Jan', 'Feb', 'March', 'April', 'May', 'June', 'July','Aug','Sep','Oct', 'Nov', 'Dec']
temp = [11, 18, 21, 35, 37, 29, 26, 31, 34, 27, 25, 18]
sales = [15, 20, 24, 37, 40, 26, 35, 34, 36, 30, 32, 25]

plt.plot(month, temp, color = 'r',
         marker = 'p', markersize = 11)
plt.plot(month, sales, color = 'g',
         marker = 'd', markersize = 11)
plt.title('tempreature varitions', fontsize = 15)
plt.xlabel('months', fontsize = 14)
plt.ylabel('Temp in celcius', fontsize = 14)

#adding the legends
# adding the legends
plt.legend(['Temp.', 'Sales'])


# grid
plt.grid(True)
plt.show()
```

add the legends to indicate the categories

```python
# create a line plot
temp = [11, 18, 21, 35, 37, 29, 26, 31, 34, 27, 25, 18]
sales = [15, 20, 24, 37, 40, 26, 35, 34, 36, 30, 32, 25]

plt.plot(month, temp, color = 'r',
         marker = 'p', markersize = 11)
plt.plot(month, sales, color = 'g',
         marker = 'd', markersize = 11)
plt.title('temperature variations', fontsize = 15)
plt.xlabel('months', fontsize = 14)
plt.ylabel('Temp in celcius', fontsize = 14)

# adding the legends
plt.legend(['Temp.', 'Sales'])
# grid
plt.grid(True)
plt.show()
```

Scatter Plots
# used to represent the data points of numerical values in the form of dots.  We want numerical data to visualize the data points and relation

```python
#creating age and weight data variations
#provide relationships between two numerical values
age = [18,20,25,19,32,35,51,29]
weight = [35,40,45,39,45,52,55,46]
```

```python
plt.scatter(age, weight, color = 'r')
plt.xlabel('Age of People', fontsize = 15)
plt.ylabel('Weight of the People', fontsize = 15)
plt.grid(open)
plt.show()
```

```python
name = ['A','B','C','D','E', 'F', 'G', 'H']
plt.scatter(name, weight, color = 'r',marker = 'p')
plt.scatter(name, age, color = 'g', marker = 'P')
plt.xlabel('Age of the person', fontsize = 15)
plt.ylabel('Weight and Name of the person', fontsize = 15)
plt.grid(open)
plt.legend('Weight', 'Age')
plt.show()
```

#Pie Chart
# - A pie chart is a circular statistical graphic divided into slices to illustrate numerical proportion, where each slice represents a part of the whole (100%). They are best used to compare 2–5 categories and show part-to-whole relationships. Key components include slices (sectors), labels, and percentages. They are easily made in Excel or Canva, but often limited by difficulty in comparing similar slice sizes.

```python
lang = ['Python', 'MySQL', 'PowerBI', 'Statistics', 'Machine L', 'AI']
fees = [500, 450, 600,700,1000,1200]
```

```python
#plot
plt.pie(fees,labels = lang)
plt.show()
```

```python
#put the % value for each category
#plot
plt.pie(fees,labels = lang, autopct = '%0.2f%%')
plt.show()
```

# highlighting the specific portion from the whole image

```python
#plot
plt.pie(fees,labels = lang, autopct = '%0.2f%%',
        explode =[0,0.1,0,0,0,0])
plt.show()
```

```python
#plot
plt.pie(fees,labels = lang, autopct = '%0.2f%%',
        explode =[0,0.2,0,0,0,0])
plt.show()
```

Subplot

```python
import matplotlib.pyplot as plt

plt.subplot(2,2, 1)
plt.plot([55,62,45,85,63,78], marker = '*')
plt.grid()

plt.subplot(2,2,2) # second plt -- bar
plt.bar(["A", 'B','C','D'], [54,25,36,44])

plt.subplot(2,2,3)#third plt -- scatter
plt.scatter([26,36,42,52], [45,52,62,59])
plt.grid(True)

plt.subplot(2,2,4) #4th --pie
plt.pie([54, 25, 36, 55], labels =['A', 'B', 'C','D'], autopct = '%0.2f%%')
plt.show()
```

A box plot (or box-and-whisker plot) summarizes data distribution using a five-number summary: minimum, first quartile (\(Q1\)), median, third quartile (\(Q3\)), and maximum. It displays the middle 50% of data within a box (interquartile range, \(IQR\)) and uses "whiskers" to show spread, with individual points outside whiskers identified as outliers.

```python
data = [ 4,6,8,10,12]

plt.boxplot(data)
plt.show()
```
